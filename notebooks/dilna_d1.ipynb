Count the placeholders

```python
(df['glucose']==0).sum()
```

```python
placeholder_cols=['glucose','blood_pressure','skin_thickness','insulin','bmi']
for c in placeholder_cols:
    print(c,'->',(df[c]==0).sum(),'zeros')
```

```python
import numpy as np
```

```python
before_missing=df.isnull().sum().sum()
```

```python
df['glucose']=df['glucose'].replace(0,np.nan)
```

```python
print(df['glucose'])
```

```python
for c in placeholder_cols:
    df[c]=df[c].replace(0,np.nan)
after_missing =df.isnull().sum().sum()
print('missing cells before:',before_missing)
print('missing cells before:',after_missing)
```

```python
df.isnull().sum()
```

```python
df['glucose']=df['glucose'].fillna(df['glucose'].median())
```

```python
print(df['glucose'])
```

```python
col_fill=['age','blood_pressure','skin_thickness','bmi','insulin']
for i in col_fill:
    df[i]=df[i].fillna(df[i].median())
    
```

```python
print(col_fill)
```

```python
df.isnull().sum()
```

```python
col_category=['smoker','physical_activity']
for i in col_category:
    df[i]=df[i].fillna('unknown')
```

```python
df[col_category].isnull().sum()
```

```python
df[col_category].isnull().sum().sum()
```
